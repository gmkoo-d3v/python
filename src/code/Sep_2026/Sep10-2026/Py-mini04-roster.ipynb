```python
# names 와 scores 가 주어집니다. 직접 시험해보려면 아래 두 줄의 # 을 지우세요.
names = ["민수", "영희", "지우"]
scores = {"민수": 88, "영희": 72, "지우": 91}

# 1단계 — 철수를 명단과 점수에 추가
names.append("철수")
scores["철수"] = 78

# 2단계 — 영희를 95점으로
scores["영희"] = 95

print(scores)
# 3단계 — 집계
average = sum(scores.values())/len(scores)
top = max(scores.values())
print(average, top, '-'*4)
passed = []
dpassed = {}

highest_value = 0
highest_person = ""
for key, value  in scores.items():
    if value >= 80:
        passed.append(key)
        dpassed[key]= value
    if highest_value < value:
        highest_value = value
        hightest_person = key

print(scores, type(scores))


print("평균", average, "/ 최고", top, "/ 통과", passed, "/ 딕셔너리", dpassed)
print(hightest_person, highest_value)
```
