명확한 이유가 있는게 아니면 global 선언해서 함수 안에서 전역 변수를 수정하는것은 하지말자. 

decorator 기존 함수를 수정하지 않고 기능을 추가할 수 있는 함수

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Start")
        result = func(*args, **kwargs)
        print("End")
        return result
    return wrapper

@decorator
def add(x,y):
    return x + y
print(add(3,5))

print('_'*50)


import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"실행시간: {end - start:.4f}초")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(1)
    print("Done")

slow_function()



```

```python
class Dog:
    name = "개똥이"
    age = 3

    def bark():
        print("왈왈")

    def run():
        print("쌩쌩")

    def let_me_introduce_myself():
        print(f"""안녕하세요. 
        나는 {Dog.name} {Dog.age}살입니다""")

print(Dog.name)
print(Dog.age)
Dog.bark()
Dog.run()
Dog.let_me_introduce_myself()

print(dir(Dog)) # 'age', 'bark', 'let_me_introduce_myself', 'name', 'run'

Dog.name = '럭키'
print(Dog.name)

```

```python
# 생성자

class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __gt__(self, target):
        return self.age > target.age

    def __add__(self,target):
        return self.age + target.age
    

    def bark(self):
        print(f'왈왈 나는 {self.name}:{self.age}살 이다')

d1 = Dog("개똥이", 3)
d2 = Dog("럭키", 5)

print(d1.__gt__(d2))
print(d1>d2)
print(d1+d2)

d1.bark()
d2.bark()
```

```python
class Human:
    population = 0
    
    def __init__(self,name):
        self.name = name
        self.age = 0
        self.hp = 0
        print("응애")
        Human.population += 1

    def aging(self):
        self.age += 1
        self.hp += 10

    def intro(self):
        print(f'{self.name}: "age"{self.age}:"hp"{self.hp}')

class NPC(Human):
    def __init__(self,name):
        super().__init__()
        self.hp = 999

    def aging(self):
        pass

h1 = Human("창민")
h2 = Human("현정")

print(Human.population)    


h1.aging()
h2.aging()

h1.intro()
h2.intro()
```

```python
class Player(Human):
    def __init__(self,name,age,weapon):
        super().__init__(name)
        self.name = name
        self.age = age
        for _ in range(age):
            self.aging()
        self.weapon = weapon
    def atk(self,target):
        target.hp -= self.weapon.dmg
        print(f'{self.name}은 {target.name}을 공격하였다')

class Weapon:
    dmg = 1

m1 = Weapon
p1 = Player("개똥이",20,m1)
p1.intro()
p1.atk(h1)
h1.intro()
```
