```python
data = ["apple", "banana", "grape", ""]

# 모두 비어 있지 않은가?
print(all(data)) # False -> '' False 로 간주
# 하나라도 비어 있는가?
print(any(not item for item in data)) # True

nums = [2,4,6,8]
# 모든 수가 짝수 인가?
print(all(n % 2 == 0 for n in nums)) # True
# 하나라도 5 이상인가?
print(any(n >= 5 for n in nums)) # True
```

### Generator

```python
# 일반
def normal():
    return [1,2,3]
# 제네레이터
def gen():
    yield 1
    yield 2
    yield 3
# Ragne 를 만들어보자
def my_range(n):
    i = 0
    while i < n:
        yield i
        i += 1

for num in my_range(5):
    print(num)
```

```python
# generator expression
g = (x ** 2 for x in range(5))
print(next(g)) # 0
print(list(g)) # [1, 4, 9, 16]
```

```python
# map filter 를 lazy 하게
nums = range(1000000)
squares = (x ** 2 for x in nums if x % 2 == 0)
for i in squares:
    if i > 100:
        break
    print(i)
```

### lambda

```python
# 람다 함수
add = lambda x, y: x + y
print(add(3, 5)) # 8

square = lambda x: x ** 2
print(square(4))

# 일반함수 
def add(x,y):
    return x + y

is_even = lambda x: "Even " if x % 2 == 0 else "Odd"
print(is_even(7)) # Odd
```
