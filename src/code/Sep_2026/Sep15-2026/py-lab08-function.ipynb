```python
def grade(score):
    """점수를 받아 등급 문자열('A'/'B'/'C'/'F')을 돌려줍니다."""
    # 여기를 완성하세요.
    if score >= 90:
        score = "A"
    elif score >= 80:
        score = "B"
    elif score >= 70:
        score = "C"
    elif score < 70:
        score = "F"   
        
    return score


```
