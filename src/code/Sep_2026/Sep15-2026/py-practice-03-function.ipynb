### 문제 1-6. 함수 안의 반복문 · 6분

1부터 `n` 까지의 합을 돌려주는 함수를 만드세요. 공식이 아니라 **반복**으로 더하세요.

**만들 함수**: `sum_to(n)` — `n` 이 0 이하면 0을 돌려줍니다

<details>
<summary>힌트 보기</summary>

누적 변수는 **함수 안에서** 0으로 시작해야 합니다. 밖에 두면 두 번째 호출부터 값이 이상해집니다.

</details>

```python
def sum_to(n):
    
    if n <= 0:
        return 0
    else:
        result = 0
        for i in range(n+1):
            result += i
        return result



print(sum_to(10), sum_to(100), sum_to(0))
```

**채점 1-6** — 아래 셀을 실행하세요.

```python
assert sum_to(10) == 55, "실제: " + repr(sum_to(10))
assert sum_to(100) == 5050, "실제: " + repr(sum_to(100))
assert sum_to(1) == 1, "실제: " + repr(sum_to(1))
assert sum_to(0) == 0, "0 이하면 0 입니다 / 실제: " + repr(sum_to(0))
assert sum_to(-5) == 0, "음수도 0 이어야 합니다 / 실제: " + repr(sum_to(-5))
print("통과")
```

평균은 소수점 **2자리**로 반올림합니다. 빈 리스트면 셋 다 0 입니다.

**만들 함수**: `stats(nums)`

<details>
<summary>힌트 보기</summary>

여러 값을 이름과 함께 돌려줘야 할 때는 튜플보다 딕셔너리가 읽기 좋습니다.

</details>

```python
def stats(nums):
        if len(nums) == 0:
                return {"count":0, "sum":0, "avg":0}
        total = 0
        for n in nums:
                total += n
        return {"count": len(nums), 'sum':total, "avg":round(total/len(nums),2)}



print(stats([1, 2, 3, 4]))
```

**채점 2-4** — 아래 셀을 실행하세요.

```python
assert stats([1, 2, 3, 4]) == {"count": 4, "sum": 10, "avg": 2.5}, "실제: " + repr(stats([1, 2, 3, 4]))
assert stats([10]) == {"count": 1, "sum": 10, "avg": 10.0}, "실제: " + repr(stats([10]))
assert stats([1, 2])["avg"] == 1.5, "실제: " + repr(stats([1, 2])["avg"])
assert stats([1, 1, 1])["avg"] == 1.0
assert stats([])["count"] == 0 and stats([])["sum"] == 0, "빈 리스트는 0 / 실제: " + repr(stats([]))
print("통과")
```

```python
def bmi(weight, height):
    value = round(weight / (height ** 2), 1)
    return value


def bmi_grade(value):
    if value <= 18.5: return '저체중'
    elif value <= 23: return '정상'
    elif value <= 25: return '과체중'
    else: return '비만'


def check(weight, height):
    # 위 두 함수를 이어서 쓰세요
    return bmi_grade(bmi(weight, height))

print(bmi(68.5, 1.72), check(68.5, 1.72))
```

```python
assert bmi(68.5, 1.72) == 23.2, "실제: " + repr(bmi(68.5, 1.72))
assert bmi(50, 1.6) == 19.5, "실제: " + repr(bmi(50, 1.6))
assert bmi_grade(17.0) == "저체중" and bmi_grade(30.0) == "비만"
assert bmi_grade(23.2) == "과체중", "실제: " + repr(bmi_grade(23.2))
assert check(68.5, 1.72) == "과체중", "실제: " + repr(check(68.5, 1.72))
assert check(50, 1.6) == "정상", "실제: " + repr(check(50, 1.6))
print("통과 — 작은 함수를 이어 붙이면 큰 일이 됩니다")
```

### 문제 5-3. 전화번호 가리기를 함수로 · 6분

가운데 자리를 `*` 로 가립니다. 이번에는 **자리 수가 달라도** 동작해야 합니다.

`"010-1234-5678"` → `"010-****-5678"`, `"02-123-4567"` → `"02-***-4567"`

**만들 함수**: `mask_phone(phone)` — 하이픈으로 세 조각이 아니면 원본을 그대로 돌려줍니다

<details>
<summary>힌트 보기</summary>

`split("-")` 로 세 조각을 얻고, 가운데 조각의 **길이만큼** 별표를 만듭니다.

</details>

```python
def mask_phone(phone):
    parts = phone.split("-")
    if len(parts) != 3:
        return phone
    return parts[0] + "-" + "*" * len(parts[1]) + "-" + parts[2]



print(mask_phone("010-1234-5678"))
print(mask_phone("02-123-4567"))
```

**채점 5-3** — 아래 셀을 실행하세요.

여기까지가 **함수**의 기본입니다. 지금 할 수 있는 일을 정리하면 이렇습니다.

- 반복되는 코드에 이름을 붙이고, 필요할 때마다 부른다
- 값을 받아(인자) 값을 돌려준다(return). 여러 값도, 없음(None)도 돌려준다
- 자주 쓰는 값을 기본값으로 두고, 필요할 때만 키워드로 바꿔 넘긴다
- 함수 안의 변수는 함수 안에서만 산다는 것을 안다
