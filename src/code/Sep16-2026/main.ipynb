filter : for 뒤에 [x for x in ... if condition] (remain with meet a condition)
값 바꾸기 ["짝" if x % 2 == 0 else '홀' for x in ...] 삼항 연산자

````python
# 딕셔너리: {key: value for ...}
squares= {x: x ** 2 for x in range(1, 6)}
print(squares) # {1:1, 2:4, 4:16, 5:25}

# 집합: {value for ...}
letters = {c for c in "banana"}
print(letters) # {'b', 'a', 'n'} 중복제거
````

````python
# 일반 함수
def squares(x):
    return x ** 2

# 같은 걸 람다로 - lambda 매개변수: 반환식
square2 = lambda x: x ** 2

pppp = lambda n: [x for x in range(n) if x < 5 if x % 2 == 0]

print(pppp(3))
print(squares(5))
print(square2(5))
````

````python
# (이름, 점수) 목록을 점수 기준으로 정렬
students = [("철수", 85),("영희", 92), ("민수", 78)]

# key: 각 항목의 무엇으로 정렬할지 (여기선 [1]=점수)
by_score = sorted(students, key=lambda s: s[1])
print(by_score)

by_name = sorted(students, key=lambda s: s[0], reverse=True)
print(by_name)

# 내림차순(높은 점수부터)
top = sorted(students, key=lambda s: s[1], reverse = True)
print(top)
````

````python
# yield로 값을 하나씩 내주는 제너레이터
def squres_gen(n):
    for i in range(n):
        yield i ** 2

# next()로 하나씩 꺼내기
gen = squres_gen(4)
print(next(gen)) # 0
print(next(gen)) # 1
print("-"*20)
# 보통은 for 로 순회 알아서 하나씩 인출
for x in squres_gen(4):
    print(x)  # 0,1,4,9
print("-"*20)
# 제네레이터 표현식 - 컴프리헨션의 소괄호
squared = (x ** 2 for x in range(5))
print(sum(squared)) # 0+1+4+9+16 =30
````

````python
scores = [
    {"name": "철수", "score": 85},
    {"name": "영희", "score": 92},
    {"name": "민수", "score": 58},
    {"name": "지영", "score": 73},
]

# 합격자(60점 이상) 이름만
passed = [s["name"] for s in scores if s["score"] >= 60]
print(f"합격: {passed}")

# 점수 높은 순 정렬 (람다 key)
ranking = sorted(scores, key=lambda s: s["score"], reverse=True)
for rank, b in enumerate(ranking, 1):
    print(f"{rank}등: {b['name']} ({b['score']}점)", end = " ")

# 평균 (합계 / 인원)
avg = sum(k["score"] for k in scores) / len(scores)
print(f"\n평균: {avg}")
````

````python
def solution(n):
    
    sum = 0
    if n % 2 == 0:
        for i in range(n):
            i += 1
            if i % 2 == 0:
                sum += i ** 2
                print('sum',sum)

    else:
        for i in range(n):
            i += 1
            if i % 2 != 0:
                sum += i
                print('sum',sum)
    return sum
              
solution(int(input()))
````
