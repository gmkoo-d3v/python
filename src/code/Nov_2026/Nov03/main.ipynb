```python
def solution(k, dungeons):
    answer = 0
    n = len(dungeons)

    # (현재 피로도, 방문한 던전 집합, 탐험 횟수)
    stack = [(k, set(), 0)]

    while stack:
        exhaustion, visited, count = stack.pop()

        answer = max(answer, count)

        for idx, (required, cost) in enumerate(dungeons):
            if idx not in visited and exhaustion >= required:
                stack.append((
                    exhaustion - cost,
                    visited | {idx},
                    count + 1
                ))

    return answer


d = [[80, 20], [50, 40], [30, 10]]
k = 80

print(solution(k, d))  # 3
```

```python
def solution(k, dungeons):
    answer = 0
    visited = [False] * len(dungeons)

    def dfs(exhaustion, count):
        nonlocal answer

        answer = max(answer, count)

        for idx, (required, cost) in enumerate(dungeons):
            if not visited[idx] and exhaustion >= required:
                visited[idx] = True

                dfs(exhaustion - cost, count + 1)

                visited[idx] = False  # 백트래킹

    dfs(k, 0)
    return answer

d = [[80, 20], [50, 40], [30, 10]]
k = 80

print(solution(k, d))  # 3
```

```python
from itertools import permutations

def solution(k, dungeons):
    answer = 0

    for order in permutations(dungeons):
        exhaustion = k
        count = 0

        for required, cost in order:
            if exhaustion >= required:
                exhaustion -= cost
                count += 1
            else:
                break

        answer = max(answer, count)

    return answer


d = [[80, 20], [50, 40], [30, 10]]
k = 80

print(solution(k, d))  # 3
```
