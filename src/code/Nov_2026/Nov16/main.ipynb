## 오차 점화식 유도

위치 갱신: $x_{t+1} = x_t + k\,e_t\,\Delta t$, 오차: $e_t = x_{target} - x_t$

양변에서 목표값을 빼면

$$e_{t+1} = e_t - k\,e_t\,\Delta t$$

$$e_{t+1} = (1 - k\Delta t)\,e_t$$

### 일반해

$$e_n = (1 - k\Delta t)^n\,e_0$$

### 수렴 조건

$n 	o \infty$ 일 때 $e_n 	o 0$ 이려면 $r^n 	o 0$ 이어야 하므로

$$\left|\frac{e_{t+1}}{e_t}\right| = |1 - k\Delta t| < 1 \iff 0 < k\Delta t < 2$$

- $0 < k\Delta t < 1$: $\dfrac{e_{t+1}}{e_t} = 1 - k\Delta t > 0$ → 부호 유지, 단조 수렴
- $1 < k\Delta t < 2$: $\dfrac{e_{t+1}}{e_t} = 1 - k\Delta t < 0$ → 부호 바뀜(오버슈트), 진동하며 수렴

### 안정성 ($r = 1 - k\Delta t$)

| $k\Delta t$ | $r$ | 거동 |
|---|---|---|
| $0 < k\Delta t < 1$ | $0 < r < 1$ | 부드럽게 수렴 |
| $k\Delta t = 1$ | $r = 0$ | 한 번에 도달 |
| $1 < k\Delta t < 2$ | $-1 < r < 0$ | 오버슈트, 진동하며 수렴 |
| $k\Delta t = 2$ | $r = -1$ | 지속 진동 |
| $k\Delta t > 2$ | $r < -1$ | 발산 |

## 해설

### 1. 무슨 상황인가

목표 위치까지 거리(오차 $e_t$)를 보고, 그 거리에 비례해서($k\,e_t$) 속도를 정해 $\Delta t$ 동안 움직이는 제어입니다. 멀수록 빨리, 가까울수록 천천히 움직입니다.

### 2. 점화식이 나오는 이유

한 스텝에 오차의 $k\Delta t$ 비율만큼을 줄입니다. 그래서 새 오차는 "옛 오차 − 줄인 만큼"입니다.

$$e_{t+1} = e_t - k\Delta t\,e_t = (1 - k\Delta t)\,e_t$$

즉 매 스텝 오차에 같은 수 $r = 1 - k\Delta t$ 를 곱합니다. 그래서 $n$ 스텝 뒤에는 $r$ 을 $n$ 번 곱한 $e_n = r^n e_0$ 가 됩니다.

### 3. $r$ 의 크기와 부호가 거동을 결정한다

- **크기 $|r|$**: 1보다 작으면 곱할수록 0에 가까워져(수렴), 1보다 크면 커집니다(발산). 그래서 수렴 조건이 $|1-k\Delta t|<1$ 이고, 풀면 $0<k\Delta t<2$ 입니다.
- **부호**: $r>0$ 이면 오차 부호가 그대로여서 목표를 넘지 않고 다가갑니다. $r<0$ 이면 매 스텝 부호가 뒤집혀 목표를 넘어갔다가 되돌아옵니다(오버슈트).

### 4. 구간별 직관

- $k\Delta t<1$: 한 스텝에 오차의 일부만 줄입니다. 목표를 넘지 않고 부드럽게 다가갑니다.
- $k\Delta t=1$: 한 스텝에 오차 전체를 줄입니다. $r=0$ 이라 바로 도달합니다.
- $1<k\Delta t<2$: 오차보다 더 많이 움직여 목표를 지나칩니다. 지나친 양은 원래 오차보다 작아서(|r|<1) 왔다 갔다 하며 점점 줄어듭니다.
- $k\Delta t=2$: 정확히 오차의 2배를 움직여 반대편 같은 거리에 도착합니다. 크기가 줄지 않아 계속 진동합니다.
- $k\Delta t>2$: 지나치는 양이 원래 오차보다 커져서 스텝마다 점점 커집니다. 발산합니다.

### 5. 실무적 의미

$k$ 가 같아도 제어 주기 $\Delta t$ 가 길어지면 $k\Delta t$ 가 커져 불안정해집니다. 예를 들어 $k=4$ 일 때 $\Delta t=0.2$ 는 $0.8$ 로 안정, $\Delta t=0.6$ 은 $2.4$ 로 발산합니다. 안정하려면 $\Delta t < 2/k$ 로 잡아야 하고, 오버슈트 없이 부드러운 응답을 원하면 $\Delta t < 1/k$ 로 잡습니다.

## 연속시간 극한 ($\Delta t 	o 0$)

### 미분방정식 유도

이산 점화식을 다시 씁니다.

$$e_{t+1} = e_t - k\,e_t\,\Delta t$$

이항하고 양변을 $\Delta t$ 로 나눕니다.

$$\frac{e_{t+1} - e_t}{\Delta t} = -k\,e_t$$

$\Delta t 	o 0$ 극한에서 좌변은 미분계수가 됩니다. 이산 인덱스를 연속 시간 $t$ 로 바꾸면

$$\frac{de(t)}{dt} = -k\,e(t)$$

### 해

일계 선형 미분방정식이므로 해는 지수함수입니다.

$$e(t) = e_0\,e^{-kt}$$

### 시간상수

지수 감소율 $k$ 의 역수 $	au = 1/k$ 를 **시간상수**라고 합니다. $t = 	au$ 일 때 오차는 초기값의 $e^{-1} \approx 36.8\%$ 로 줄어듭니다.

| 시간 | 오차 비율 |
|---|---|
| $t = 	au = 1/k$ | $e^{-1} \approx 36.8\%$ |
| $t = 2	au$ | $e^{-2} \approx 13.5\%$ |
| $t = 3	au$ | $e^{-3} \approx 5.0\%$ |
| $t = 4	au$ | $e^{-4} \approx 1.8\%$ |
| $t = 5	au$ | $e^{-5} \approx 0.7\%$ |

일반적으로 $5	au$ 이후에는 오차가 거의 0으로 간다고 봅니다.

### 이산 vs 연속 비교

이산 해 $e_n = (1 - k\Delta t)^n\,e_0$ 에서 $n = t/\Delta t$ 이므로

$$e(t) = (1 - k\Delta t)^{t/\Delta t}\,e_0$$

$\Delta t 	o 0$ 극한을 계산하면

$$\lim_{\Delta t 	o 0}(1 - k\Delta t)^{t/\Delta t} = e^{-kt}$$

연속 해 $e_0\,e^{-kt}$ 와 일치합니다. $\Delta t$ 가 작을수록 이산 시스템은 연속 시스템을 잘 근사합니다.

### 해설: 왜 지수함수인가

매 순간 오차에 비례해서 줄이므로, 줄어드는 속도 자체가 오차에 비례합니다. 오차가 클 때는 빠르게, 작을 때는 느리게 줄어듭니다. "변화율이 현재값에 비례한다"는 것은 지수함수의 정의와 같습니다. 연속시간에서는 $\Delta t 	o 0$ 이므로 $r = 1-k\Delta t$ 가 항상 $1$ 에 가까워 부호가 음수가 될 수 없습니다. 따라서 연속시간에서는 오버슈트나 발산이 없고, 항상 부드럽게 수렴합니다. 오버슈트나 발산은 $\Delta t$ 가 충분히 클 때만 나타나는 **이산시간의 고유 현상**입니다.
