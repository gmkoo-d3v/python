```python
from collections import deque

def solution(numbers, target):
    answer = 0

    graph = deque()
    graph.append((0,0))
    while graph:
        idx, tango = graph.popleft()
        if idx == len(numbers):
            if tango == target:
                answer += 1
            continue
        graph.append((idx+1, tango+numbers[idx]))
        graph.append((idx+1, tango-numbers[idx]))
    return answer


n = [4,1,2,1]
t = 4
solution(n,t)
```

```python
def solution(numbers, target):
    answer = 0

    def dfs(index, total):
        nonlocal answer
        if index == len(numbers):
            if total == target:
                answer += 1
            return
        dfs(index + 1, total + numbers[index])
        dfs(index + 1, total - numbers[index])
    dfs(0,0)
    return answer

n = [4,1,2,1]
t = 4
solution(n,t)
```
