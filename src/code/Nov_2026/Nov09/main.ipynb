```python
def solution(maps):
    rows, cols = len(maps), len(maps[0])
    visited = [[False] * cols for _ in range(rows)]

    answer = []

    def dfs(r, c):
        stack = [(r, c)]
        visited[r][c] = True
        total = 0

        while stack:
            r, c = stack.pop()
            total += int(maps[r][c])

            for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
                nr, nc = r + dr, c + dc

                if 0 <= nr < rows and 0 <= nc < cols:
                    if maps[nr][nc] != 'X' and not visited[nr][nc]:
                        visited[nr][nc] = True
                        stack.append((nr, nc))

        return total

    for r in range(rows):
        for c in range(cols):
            if maps[r][c] != 'X' and not visited[r][c]:
                answer.append(dfs(r, c))

    return sorted(answer) if answer else [-1]


maps = ["X591X", "X1X5X", "X231X", "1XXX1"]
maps1 = ["XXX", "XXX", "XXX"]

print(solution(maps))   # [1, 1, 27]
print(solution(maps1))  # [-1]
```

```python
from collections import deque

def solution(maps):
    rows, cols = len(maps), len(maps[0])
    visited = [[False] * cols for _ in range(rows)]
    answer = []

    for r in range(rows):
        for c in range(cols):
            if maps[r][c] != 'X' and not visited[r][c]:
                visited[r][c] = True
                q = deque([(r, c)])
                total = 0

                while q:
                    cr, cc = q.popleft()
                    total += int(maps[cr][cc])

                    for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
                        nr, nc = cr + dr, cc + dc

                        if 0 <= nr < rows and 0 <= nc < cols:
                            if maps[nr][nc] != 'X' and not visited[nr][nc]:
                                visited[nr][nc] = True
                                q.append((nr, nc))

                answer.append(total)

    if not answer:
        return [-1]

    return sorted(answer)


maps = ["X591X", "X1X5X", "X231X", "1XXX1"]
maps1 = ["XXX", "XXX", "XXX"]

print(solution(maps))   # [1, 1, 27]
print(solution(maps1))  # [-1]
```
