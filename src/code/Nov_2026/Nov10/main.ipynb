```python
def solution(nums):
    num = {}
    total = len(nums) // 2
    count = 0

    for i in nums:
        num[i] = num.get(i, 0) + 1

    while total > 0 and num:
        i = next(iter(num))

        count += 1
        total -= 1
        del num[i]

    return count

n = [3,3,3,2,2,4]
n1 = [3,3,3,2,2,2]
n2 = [3,1,2,3]
solution(n2)
solution(n)
solution(n1)
```
