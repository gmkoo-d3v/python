```python
def solution(n, computers):
    visited = [False] * n
    answer = 0

    def dfs(node):
        visited[node] = True
        for next_node in range(n):
            # 연결되어 있고 아직 방문하지 않은 컴퓨터라면 탐색 계속
            if computers[node][next_node] == 1 and not visited[next_node]:
                dfs(next_node)

    for i in range(n):
        if not visited[i]:
            dfs(i)          # 하나의 네트워크에 속한 모든 컴퓨터 방문
            answer += 1     # 네트워크 개수 증가

    return answer
    
n = 3
c = [[1, 1, 0], [1, 1, 0], [0, 0, 1]]
c2 = [[1, 1, 0], [1, 1, 1], [0, 1, 1]]
solution(n,c)
solution(n,c2)
```
