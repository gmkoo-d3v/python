```python
# 리스트의 중복을 set으로 제거하는 흔한 패턴
nums = [1, 2, 2, 3, 3, 3]
unique = set(nums)
print(unique)          # {1, 2, 3} — 중복 사라짐

s = {1, 2, 3}
s.add(4)             # 요소 추가
s.update([5, 6])     # 여러 개 추가
s.remove(1)          # 요소 삭제
print(s)
```

```python
a = {1, 2, 3, 4, 5}
b = {3, 5, 7, 9}

print(a | b)   # 합집합 → {1,2,3,4,5,7,9}
print(a & b)   # 교집합 → {3, 5}
print(a - b)   # 차집합 → {1, 2, 4}
print(a ^ b)   # 대칭차집합(양쪽 중 한쪽에만) → {1,2,4,7,9}
```

```python
import time

def prime_printer_1(n):
    start = time.time()
    prime_list = []
    for i in range(2, n+1):
        detector = 0
        for k in prime_list:
            if i % k == 0:
                detector += 1
        if detector == 0:
            prime_list.append(i)
    end = time.time()
    return end-start

print('required %10.12f seconds' %prime_printer_1(100))
print('required %10.12f seconds' %prime_printer_1(1000))
print('required %10.12f seconds' %prime_printer_1(10000))
print('required %10.12f seconds' %prime_printer_1(100000))
```

```python
import time

def prime_printer_2(n):
    start = time.time()
    prime_list = []
    for i in range(1, n+1):
        detector = 0
        for k in range(2,i):
            if i % k == 0:
                detector += 1
        if detector == 0:
            prime_list.append(i)
    end = time.time()
    return end-start

print(f'required {prime_printer_2(100):0.12f} seconds')
print('required %10.12f seconds' %prime_printer_2(1000))
print('required %10.12f seconds' %prime_printer_2(10000))
print('required %10.12f seconds' %prime_printer_2(100000))
```

```python
# 버블 정렬

array = [9,8,7,6,5,4,3,2,1]

def bubble_sort(array):
    n = len(array)
    for i in range(n-1):
        for j in range(n - i -1):
            if array[j] > array[j + 1]:
                array[j], array[j+1] = array[j+1], array[j]
        print(array)

print("before:", array)
bubble_sort(array)
print("after:", array)
```

```python
# selection
array = [8,4,6,2,9,1,3,7,5]

def selection_sort(array):
    n = len(array)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if array[j] < array[min_index]:
                min_index =j
        array[i], array[min_index] = array[min_index], array[i]
        print(array[:i+1])

print("before: ", array)
selection_sort(array)
print("after: ", array)
```

```python
# insert
array = [8,4,6,2,9,1,3,7,5]

def insertion_sort(array):
    n = len(array)
    for i in range(1, n):
        for j in range(i, 0, -1):
            if array[j-1] > array[j]:
                array[j-1], array[j] = array[j], array[j-1]
        print(array[:i+1])

print("before: ", array)
insertion_sort(array)
print("after: ", array)
```

```python
# merger
array = [8,7,6,2,9,1,3,7,5]

def merge_sort(array):
    if len(array) < 2:
        return array
    mid = len(array) // 2
    low_arr = merge_sort(array[:mid])
    high_arr = merge_sort(array[mid:])

    merged_arr = []
    l = h = 0
    while l < len(low_arr) and h < len(high_arr):
        if low_arr[l] < high_arr[h]:
            merged_arr.append(low_arr[l])
            l += 1
        else:
            merged_arr.append(high_arr[h])
            h += 1

    merged_arr += low_arr[l:]
    merged_arr += high_arr[h:]
    print(merged_arr)
    return merged_arr

print("before: ", array)
array = merge_sort(array)
print("after: ", array)
```

```python
# quick_sort

array = [8,4,6,2,5,1,3,7,9]

def quick_sort(array):
    if len(array) <= 1:
        return array
    pivot = len(array) // 2
    front_arr, pivot_arr, back_arr = [],[],[]
    for value in array:
        if value < array[pivot]:
            front_arr.append(value)
        elif value > array[pivot]:
            back_arr.append(value)
        else:
            pivot_arr.append(value)
    print(front_arr, pivot_arr, back_arr)
    return quick_sort(front_arr) + quick_sort(pivot_arr) + quick_sort(back_arr)

print("before: ", array)
array = quick_sort(array)
print("after: ", array)
```
