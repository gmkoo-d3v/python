```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        print(f'{self.name}: 멍멍!')

# 인스턴스 생성 (붕어빵 굽기)
d = Dog("바둑이", 3)
print(d.name) # 바둑이(속성 접근)
print(d.age)  # 3
d.bark()      # 바둑이 : 멍멍! (매서드 호출)
```

self는 인스턴스 자기 자신을 가르킨다. d.bark()를 부르면, 파이썬이 자동으로 d 를 self 자리에 넣어준다. </br> self.name이 강아지 이름이 된다.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_info(self):
        return f"{self.name}, {self.age}살"

    def birthday(self):        # 나이 +1 (자기 속성 변경)
        self.age = self.age + 1
        print(f"{self.name} 생일! 이제 {self.age}살")

d = Dog("바둑이", 3)
print(d.get_info())   # 바둑이, 3살
d.birthday()            # 바둑이 생일! 이제 4살
d.age = 10            # 밖에서 직접 바꾸기도 가능
print(d.get_info())   # 바둑이, 10살
```

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance      # 기본값 0 (8장 기본값 인자)

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"{amount}원 입금 → 잔액 {self.balance}원")

    def withdraw(self, amount):
        if amount > self.balance:      # 6장 조건문
            print("잔액 부족!")
        else:
            self.balance = self.balance - amount
            print(f"{amount}원 출금 → 잔액 {self.balance}원")

# 사용
acc = Account("철수", 1000)
acc.deposit(500)     # 1500
acc.withdraw(2000)   # 잔액 부족!
acc.withdraw(300)    # 1200
```

```python
def solution(num_list):
    count = 0
    for i in range(len(num_list)):
        while num_list[i] != 1:
            
            if num_list[i] % 2 == 0:
                num_list[i] = num_list[i]/2
                
            else:
                num_list[i] = (num_list[i]-1)/2
            count +=1
    
    return count
a =[12, 4, 15, 1, 14]
print(len(a))
solution(a)
```
