````python
class Account:
    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def withdraw(self, amount):
        if amount <= self._balance:
            self._balance -= amount
            print(f"{amount}원 출금 완료")
        else:
            print(f"잔액이 부족합니다. (현재 잔고: {self._balance}원 / 요청 금액: {amount}원)")
            

    def get_balance(self):
        return self._balance

acc = Account(1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())
acc.withdraw(2000)
print(acc.get_balance())
````

````python
class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        return 0
    def __str__(self):
        return f"{self.name}의 넓이: {self.area()}"

class Rectangle(Shape):
    def __init__(self, w, h):
        super().__init__("직사각형")
        self.w = w
        self.h = h
    def area(self):          # 오버라이딩
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        super().__init__("원")
        self.r = r
    def area(self):
        return 3.14 * self.r ** 2

class Triangle(Shape):
    def __init__(self,w,h):
        super().__init__('삼각형')
        self.w = w
        self.h = h
    def area(self):
        return int(self.w * self.h /2)

# __str__ 덕분에 print가 깔끔
print(Rectangle(4, 5))   # 직사각형의 넓이: 20
print(Circle(3))        # 원의 넓이: 28.26
print(Triangle(3,4))
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def __str__(self):
        return f'동물 {self.name}'          # "동물 흰둥이"


class Dog(Animal):
    def speak(self):
        return '멍멍'


class Cat(Animal):
    def speak(self):
        return '야옹'


class Puppy(Dog):
    def __init__(self, name, age):
        super().__init__(name)                 # 부모의 준비를 먼저
        self.age = age

    def speak(self):
        return f'{super().speak()}!'          # 부모의 소리 뒤에 "!"

def chorus(animals):
    """동물들의 소리를 공백으로 이어 붙인다."""
    return " ".join(a.speak() for a in animals)
 
````

````python
# inheritance
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def hello(self):
        print(f"HI 나는 {self.name}이다. {self.age}살 이다.")

    def dance(self):
        print(f'{self.name}는 춤추기')

class NPC(Human):
    def __init__(self, name, age, mp):
        super().__init__(name, age)
        print('야호')
        self.mp = mp

    def hello(self):
        print(f'hi 나는 {self.name}이다. {self.age}살이다 나는 NPC이다')

    def cast(self):
        print(f'{self.name}은 화염구를 영창했다. 마나가 20줄었다 현재 마나는 {self.mp-20}')

h1 = Human('개똥', 30)
h1.hello()
h2 = NPC('소똥', 40, 30)
h2.hello()
h2.dance()
h2.cast()
````

````python
class Adder:
    def __init__(self,v):
        self.__v = v

    # getter: 속성을 읽을 때 호출됩니다.
    @property
    def v(self):
        return self.__v

    # setter : 속성에 값을 대입할 때(=) 호출됩니다.
    @v.setter
    def v(self, value):
        if value < 0:
            raise ValueError('Value는 음수가 될 수 없습니다.')
        self.__v = value

    def __add__(self,target):
        if isinstance(target, Adder):
            return self.v + target.v
        else: return self.v + target

    def __radd__(self,target):
        return self.__add__(target)  
    
    def hello(self):
        print(self.__v)

a = Adder(1)
b = Adder(2)
c = Adder(3)
print(type(a+b))
print(a+b)
````
