```python
def single(value):
    """값 하나만 담은 튜플. (3) 은 튜플이 아니다 — 쉼표가 튜플을 만든다."""
    return (value,)


def swap(a, b):
    """두 값을 맞바꾼다. 임시 변수 없이 한 줄."""   
    return b,a


def first_and_rest(items):
    """(첫 값, 나머지 리스트). * 를 쓰면 한 줄."""
    first, *rest = items
    return first, rest


def count_and_find(items, target):
    """(몇 개, 첫 위치). 없으면 (0, -1)."""
    if target not in items:
        return 0, -1
    return items.count(target), items.index(target)


def try_change(items):
    """튜플의 첫 원소를 바꿔본다. 실패하면 '바꿀 수 없음'."""
    try:
        items[0] = 1
    except TypeError:
        return "바꿀 수 없음"
```
