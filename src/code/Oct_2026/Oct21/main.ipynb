```python
from collections import Counter

def solution(want, number, d):
    target = dict(zip(want, number))

    answer = 0
  
    for i in range(len(d) -9): # 10일 구간의 시작일
        count = {}
        for item in d[i:i+10]:
            count[item] = count.get(item, 0) + 1
        if count == target:
            answer += 1
    return answer

want = ["banana", "apple", "rice", "pork", "pot"]
number = [3, 2, 2, 2, 1]
discount = ["chicken", "apple", "apple", "banana", "rice", "apple", "pork", "banana", "pork", "rice", "pot", "banana", "apple", "banana"]
solution(want,number, discount)
```

```python
import math
sig = lambda x: 1 / (1 + math.exp(-x))

X = [(0,0), (0,1), (1,0), (1,1)]
Y = [0, 1, 1, 0]
lr = 0.5
w1=w2=w3=w4=w5=w6 = 0.5      # 가중치 초기화는 반복 시작 전 1번

for epoch in range(1, 3001):
    for (x1, x2), y in zip(X, Y):

        # 순전파
        h1 = sig(x1 * w1 + x2 * w3)
        h2 = sig(x1 * w2 + x2 * w4)
        o  = sig(h1 * w5 + h2 * w6)      # x5 -> h2

        # 역전파
        d3 = o - y
        d1 = d3 * w5 * h1 * (1 - h1)     # w4 -> w5 (h1쪽 은닉 가중치)
        d2 = d3 * w6 * h2 * (1 - h2)

        # update
        w5 -= lr * d3 * h1
        w6 -= lr * d3 * h2
        w1 -= lr * d1 * x1
        w3 -= lr * d1 * x2               # x1 -> x2
        w2 -= lr * d2 * x1
        w4 -= lr * d2 * x2

# 학습 결과 확인 (XOR)
for (x1, x2), y in zip(X, Y):
    h1 = sig(x1 * w1 + x2 * w3)
    h2 = sig(x1 * w2 + x2 * w4)
    o  = sig(h1 * w5 + h2 * w6)
    print(f"({x1},{x2}) -> {o:.3f} (정답 {y})")
```
