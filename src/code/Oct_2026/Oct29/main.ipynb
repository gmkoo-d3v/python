```python
def solution(s):
    l, r = 0, 0
    if s[0] == chr(40) and s[-1] == chr(41):   
        for i in s:
            if i == chr(40):
                l += 1
            elif i == chr(41):
                r += 1
        if l == r:
            return True
        else: return False        
    else:
        return False
    
```

```python
def solution(s):
    answer = True
    stack = []

    for ch in s:
        if ch == '(':
            stack.append(ch)
        elif ch == ')':
            if stack:
                stack.pop()
            else:
                return False

    if stack:
        answer = False

    return answer
```

```python
def solution(s):
    stack = []

    for ch in s:
        if ch == '(':
            stack.append(ch)
        else:
            if not stack:
                return False
            stack.pop()

    return not stack
```

```python
from collections import deque

def solution(s):
    que = deque([])
    count = 0
    for idx in range(len(s)):
        que.append(s[idx])

    for i in range(len(s)):
        a = que.popleft() 
        if a == chr(40):
            count += 1
        elif count != 0 and a == chr(41):
            count -= 1
        else: return False
    if count == 0: return True
    else: return False
```

```python
from collections import deque

def solution(priorities, location):
    x = len(priorities)
    que = deque([])
    idx = deque([])
    count = 0

    for n in range(x):
        idx.append(n)

    for k in range(x):
        que.append(priorities[k])

    while que:
        a, aa = que.popleft(), idx.popleft()

        if que and a < max(que):
            que.append(a)
            idx.append(aa)
        else:
            count += 1

            if aa == location:
                return count

    return count



p = [2,1,3,2]
s = 2
solution(p,s)
```
