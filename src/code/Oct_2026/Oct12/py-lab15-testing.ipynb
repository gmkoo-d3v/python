```python

# from solution import grade, cases, invalid_inputs
# 지금까지 테스트는 **받기만** 했습니다. 이번에는 씁니다.
#
# 채점 방식이 다릅니다. 여러분이 낸 테스트를 **일부러 망가뜨린 구현들**에 돌려서,
# 그 버그를 잡아내는지 봅니다. 잡으면 통과입니다.
#
# 즉 "테스트가 통과하는가" 가 아니라 **"테스트가 버그를 잡는가"** 를 봅니다.


def grade(score):
    """점수를 등급으로. 이 함수가 시험 대상입니다.

    90 이상 A · 80 이상 B · 70 이상 C · 그 아래 F
    점수는 0~100 입니다. 벗어나면 ValueError.
    """
    if not 0 <= score <= 100:
        raise ValueError("점수는 0~100 입니다")
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    return "F"


def cases():
    """(입력, 기대하는 등급) 짝을 리스트로 돌려주세요.

    아래 버그들을 **모두** 잡아내야 합니다.
      ① 경계를 > 로 쓴 것      (90 을 B 라고 답한다)
      ② C 를 빠뜨린 것         (70~79 를 F 라고 답한다)
      ③ 순서를 뒤집은 것       (95 를 C 라고 답한다)

    잡으려면 **경계값**을 넣어야 합니다. 85 같은 가운데 값만으로는 ① 을 못 잡습니다.
    """
    return [
        (90, "A"),
        (75, "B"),
        (89, "A"),
        (80, "B"),
        (90, "C"),

    ]

def invalid_inputs():
    """ValueError 가 나야 하는 입력을 리스트로 돌려주세요.

    범위를 벗어난 값입니다. 양쪽 끝을 다 봐야 합니다.
    """
    return [
        -1,
        101,
        -100,
        200,
    ]
```
