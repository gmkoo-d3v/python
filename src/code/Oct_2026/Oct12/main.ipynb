```python
def add(a, b):
    return a + b

# input -> check pair of suspected.
assert add(2, 3) == 5
assert add(0, 0) == 0
assert add(-1, 1) == 0
assert add(100, 200) == 300
print('add 함수: 4개 케이스 모두 통과 ✓')
```

```python
def grade(score):
    if score < 0 or score > 100:
        return '점수가 아닙니다.'
    elif score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "F"

# 정상값
assert grade(95) == 'A'
assert grade(85) == 'B'
# 경계값 - 딱 90, 딱 80, 그리고 한 끗 아래 89
assert grade(90) == "A"
assert grade(89) == "B"
assert grade(80) == "B"
# 0, 음수 같은 극단 값
assert grade(0) == 'F'
assert grade(-5) == "점수가 아닙니다."
print("정상·경계·극단까지 모두 통과 ✓")
```

```python
def solution(s, skip, index):
    all_list = []
    answer =''

    for ch in range(97, 97+26):
        if chr(ch) in skip:
            continue
        all_list.append(chr(ch))

        list_len = len(all_list)
    for i in range(len(s)):
        ind_all = all_list.index(s[i])
        answer += all_list[(ind_all+index) % list_len]
    return answer

print(ord('a'))
s = 'aukks'
skip = 'wbqd'
index = 5
solution(s, skip, index)
```
