```python
def solution(N, number):
    if N == number:
        return 1

    dp = [set() for _ in range(9)]         # dp[0]is not use

    repeat = 0 
    for i in range(1, 9):
        repeat = repeat * 10 + N           # 5 -> 55 -> 555 ...
        dp[i].add(repeat)

        for j in range(1, i):              # 2. dp[j] @ dp [i-j]
            # cross 로 사칙연산
            for a in dp[j]:
                for b in dp[i-j]:
                    dp[i].add(a+b)
                    dp[i].add(a-b)
                    dp[i].add(a*b)
                    if b != 0:
                        dp[i].add(a//b)

        if number in dp[i]:
            print(dp)
            return i
    return -1
print(solution(5,12))
```
