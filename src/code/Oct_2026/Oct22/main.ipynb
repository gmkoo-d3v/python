```python
def solution(want, number, d):
    target = []
    for item, num in zip(want, number):
        target += [item] * num

    target.sort()

    answer = 0
    for i in range(len(d) -9):
        if sorted(d[i:i+10]) == target:
            answer += 1
    return answer

want = ["banana", "apple", "rice", "pork", "pot"]
number = [3, 2, 2, 2, 1]
discount = ["chicken", "apple", "apple", "banana", "rice", "apple", "pork", "banana", "pork", "rice", "pot", "banana", "apple", "banana"]
solution(want,number, discount)
```

```python
def solution(pb):
    pb.sort()

    for idx, phone in enumerate(pb[:-1]):
        if pb[idx + 1].startswith(phone):
            return False

    return True

# pb1 = ["119", "97674223", "1195524421"]
# solution(pb1)
# pb2 = ["123","456","789"]
# solution(pb2)
pb3 = ["12","123","1235","567","88"]
solution(pb3)
```
