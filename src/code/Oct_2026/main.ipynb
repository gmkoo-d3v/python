```python
from collections import deque

def solution(b, w, tw):
    answer = 0
    bridge = deque([0] * b) # 0 은 빈칸
    trucks = deque(tw)
    total = 0 # 다리 위 트럭 무게 합

    while trucks:
        answer += 1

        # 맨 앞 칸이 다리에서 내려감
        total -= bridge.popleft()

        # 다음 트럭이 올라갈 수 있으면 올리고, 아니면 빈 칸 추가
        if total + trucks[0] <= w:
            truck = trucks.popleft()
            total += truck
            bridge.append(truck)
        else:
            bridge.append(0)
    # 마지막 트럭이 다리를 다 건너는 시간
    answer += b
    return answer

bl = 2
w = 10
tw=	[7,4,5,6]

solution(bl,w,tw)
```
