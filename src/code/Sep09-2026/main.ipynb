```python
# 리스트의 문제점
position = [10, 20]
print(f"원래 좌표: {position}")

# 실수로 변경될 수 있습니다!
position[0] = 100
print(f"변경된 좌표: {position}")  # 좌표가 바뀌어버렸습니다

# 튜플로 해결
position = (10, 20)  # 소괄호 () 사용
print(f"좌표: {position}")

# 변경 시도
try:
    position[0] = 100
except TypeError as e:
    print(f"❌ 에러: {e}")
    print("→ 튜플은 변경할 수 없습니다!")
```

```python
# copy
x = [1,2,3]
print(x,id(x),'x')
print('-------------')
# y=x # 같은 메모리 주소가됨
# shallow copy
y = x.copy() #메모리 값은 같지 않다.
print(x,id(x),'x')
print(y,id(y),'y')
print('-------------')

x = 99
print(x,id(x),'x')
print(y,id(y),'y')
```

```python
# shallowed copy
x = [[1,2],3,4]
print(x,id(x),'x')
print('-------------')
y = x.copy()
y[1] = -1121
print(x,id(x),'x')
print(y,id(y),'y')
print('-------------')

y[0][0] = 99
print(x,id(x),'x')
print(y,id(y),'y')
print(id(x[0]),id(y[0]))
print(x[0][1],y[0][1])
print(x[1],y[1])
```

```python
# deep copy
import copy
x = [[1,2],3,4]
y = copy.deepcopy(x)
print(id(x[0]),id(y[0]))
y[0][0] = 99
print(x)
print(y)
print(id(x[0]),id(y[0]))
# 가변 객체 안에 가변객체가 있다면 copy.deepcopy해야 된다.
```
