```python
# raw 가 주어집니다. 직접 시험해보려면 아래 줄의 # 을 지우세요.
raw = "  Hong.GilDong@Example.COM  "
raw = "\tKim@Test.co.kr\n"

# 1단계 — 앞뒤 공백 없애고 소문자로
cleaned = raw.lower().strip()

print(cleaned)
# 2단계 — @ 앞뒤로 나누기
user, domain = cleaned.split('@')


# 3단계 — 보고 문자열
report = f'{user} / {domain}'

print(report)
```
