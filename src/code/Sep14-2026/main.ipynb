````python
def str_spacer(input_str:str)->str:
    try:
        out_str = ""
        for single_char in input_str:
            out_str += single_char + " "
        return out_str[:-1] # 5뒤에 스페이스 붙는걸 제거하기 위해서 [:] 해버리면 len(spaced_out) 값이 10이 된다.
    except:
        print("input must be string")
        return None
    else:
        print("else!!")
    finally:
        print("fianlly: try에서 return 되지만 이것은 출력되네?!")

spaced_out = str_spacer('12345')
print(spaced_out)
print(len(spaced_out))

print(str_spacer.__annotations__)    
````

````python
def calculator(x):
    def operate(y, op = '+'):
        if op == '+':
            return x + y
        elif op == '-':
            return x-y
        elif op == '*':
            return x*y
        elif op == '/':
            return x/y
        else:
            return '지원하지 않는 연산입니다.'
    return operate
f = calculator(10)
print(f(5))
print(f(3,'*'))
print(f(2,'-'))
print(f(4,'/'))

print('-'* 10)
def power_base(base):
    def raise_to_power(exp):
        return base ** exp
    return raise_to_power

f = power_base(3)
print(f(2))
print(f(4))

print('-'* 10)

square = power_base(2)
cube = power_base(3)
print(square(5))
print(cube(2))

print('-'* 10)
def make_coupon(discoutn_rate):
    def apply_discount(price):
        return price * (1 - discoutn_rate)
    return apply_discount


coupon10 = make_coupon(0.10)
coupon25 = make_coupon(0.25)
print(coupon10(10000))
print(coupon25(10000))
````

decorator

````python
def decorator(func):
    def wrapper(*args, **kwargs):
        print('start')
        result = func(*args, **kwargs)
        print('END')
        return result
    return wrapper

@decorator
def add(x,y):
    return x+ y
print(add(3,5))
````

````python
import time
def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f'실행 시간: {end - start:.4f}초')
        return result
    return wrapper
@timer
def slow_function():
    time.sleep(1)
    print('Done')    
slow_function()
````

### 하노이의 탑

````python
def hanoi(num, _from, _by, _to):
    # escape condition
    if num == 1:
        print("{}에서 {}로 {}번째 원반이동".format(_from, _to, num)) 
        print('xxxxx')       
        return
    print(f'num of tray:{num} start')
    hanoi(num-1, _from, _to, _by)
    print("{}에서 {}로 {}번째 원반이동".format(_from, _to, num))
    hanoi(num-1, _by, _from, _to)
    print(f'num of tray:{num} last line')

while 1:
    numOfTray = int(input('원반의 개수를 입력하세요!(종료:0):'))
    print('-'*10)
    print(f'input:{numOfTray}')
    if numOfTray == 0:
        break
    hanoi(numOfTray, 'A', 'B', 'C')
````
