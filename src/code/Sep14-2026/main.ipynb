````python
total = 0
for i in range(1, 101):
    if i % 2 == 0:
        total = total + i
print(f'1~100 짝수의 합: {total}') # 2550
totalodd = 0
for j in range(1, 101):
    if j % 2 != 0: # if not j %2 == 0:
        totalodd += j
print(f'1~100 홀수의 합: {totalodd}')
````

````python
answer = 7
while True:
    guess = int(input('1~10 중 맞춰보세요: '))
    if guess == answer:
        print('정답')
        break
    elif guess < answer:
        print("더 큰 수에요.")
    else:
        print('더 작은 수 예요.')
````
