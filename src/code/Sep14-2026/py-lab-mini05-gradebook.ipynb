```python
# scores 가 주어집니다. 직접 시험해보려면 아래 줄의 # 을 지우세요.
scores = [95, 83,  99]

grades = []
counts = {"A": 0, "B": 0, "C": 0, "F": 0}
highest = scores[0]
lowest = scores[0]
total = 0

for score in scores:
    if score >= 90:
        grade = 'A'
    elif score >=80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    else:
        grade = 'F'
    grades.append(grade)
average = round(sum(scores)/len(scores), 1)

for grade in grades:
    if grade == 'A':
        counts['A'] += 1
    elif grade == 'B':
        counts['B'] += 1
    elif grade == 'C':
        counts['C'] += 1
    else:
        counts['F'] += 1
for i in scores:
    if i > highest:
        highest = i
    if i < lowest:
        lowest = i

print(f"A {counts['A']}")
print(f"B {counts['B']}")
print(f"C {counts['C']}")
print(f"F {counts['F']}")
```
