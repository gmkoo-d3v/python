```python
import math
from matplotlib import pyplot as plt
def add(v,w):
    ''' Adds corresponding elements'''
    assert len(v) == len(w), "Vectors must be the same length"

    return [vi + wi for vi, wi in zip(v, w)]



v = [1,2]
w = [4,5]

add (v,w)
def plot(v,w):
    from matplotlib import pyplot as plt
    plt.xlim(0,10)
    plt.ylim(0,10)

    plt.scatter(v,w)
    plt.show()

plot(v,w)
```

```python
import numpy as np
from matplotlib import pyplot as plt

A = np.array([[1,2],[2,5]])
B = np.array([4,9])

solution = np.linalg.solve(A,B)
print("연립방정식의 해:", solution)
```
