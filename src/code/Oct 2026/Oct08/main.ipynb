```python
# -*- coding: utf-8 -*-
def add(path, memo):
    """메모 한 줄을 파일 뒤에 덧붙입니다."""
    
    with open(path, 'a') as f:
        f.write(memo + "\n")


def read_all(path):
    """모든 메모를 리스트로. 파일이 없으면 빈 리스트."""
    memos = []
    try:
        with open(path,'r') as f:
            memos = f.readline()
    except FileNotFoundError:
        print('file is not found')
    return memos

def find(path, word):
    """그 단어가 들어간 메모만 골라 돌려줍니다."""
    memos = read_all(path)
    return [memo for memo in memos if word in memo]
            
            
    
def clear(path):
    """메모를 전부 지웁니다."""
    with open(path,'w') as f:
        f.write("")
```

```python
def safe_get(data, index):
    try:
        return data[index]
    except IndexError:
        return "범위를 벗어났어요"
    except TypeError:
        return "인덱스는 숫자여야 해요"

nums = [10, 20, 30]
print(safe_get(nums, 1))      # 20
print(safe_get(nums, 99))     # 범위를 벗어났어요
print(safe_get(nums, "a"))    # 인덱스는 숫자여야 해요
```

```python
def divide(a,b):
    try:
        result = a / b
    except ZeroDivisionError:
        print('0으로 나눌 수 없어요.')
    else:
        print(f'결과: {result}') # 성공 시만
    finally:
        print('계산 완료')

divide(10, 2)
print('_'* 10)
divide(10,0)
```

```python
import math


def circle_area(r):
    """원의 넓이. 소수 둘째 자리에서 반올림."""
    # result = math.pi * math.sqrt(r)

    # return round(result,1)
    return round(math.pi * r ** 2 ,2)

def safe_divide(a, b):
    """0으로 나누면 None. 죽지 않는다."""
    try:
        result = a / b
    except ZeroDivisionError:
        return None
    return result


def parse_number(text):
    """정수로 바꾼다. 숫자가 아니면 None."""
    try:
        # result = int(text)
        # return result
        return int(text)
    except ValueError:
        return None
    


def classify_error(fn):
    """fn 을 실행하고 난 에러의 이름을 돌려준다. 성공하면 '성공'."""
    try:
        fn()
        return "성공"
    except ZeroDivisionError:
        return "ZeroDivisionError"
    except ValueError:
        return "ValueError"
    except TypeError:
        return "TypeError"
    except KeyError:
        return "KeyError"
    except Exception as e:
        # return f'알 수 없는 에러 {type(e).__name__}'
        return '알 수 없는 에러'


def with_cleanup(fn, log):
    """성공하든 실패하든 log 에 '정리' 를 넣는다."""
    try:
        result = fn()
    except Exception:
        return None
    else:
        return result
    finally:
        log.append('정리')
```

```python
import re


def parse(line):
    match = re.match(
        r"(\d{4}-\d{2}-\d{2}) (\d{2}:\d{2}) \[(\w+)\] (.*)",
        line
    )

    if match:
        return match.groups()

    return None


def count_levels(lines):
    """등급별 건수. 깨진 줄은 세지 않습니다."""

    level_counts = {}

    for line in lines:
        parsed = parse(line)

        if parsed:
            _, _, level, _ = parsed
            level_counts[level] = level_counts.get(level, 0) + 1

    return level_counts

def busiest_hour(lines):
    """가장 많은 로그가 찍힌 시(두 자리 문자열). 없으면 None."""

    hour_counts = {}

    for line in lines:
        parsed = parse(line)

        if parsed:
            _, time, _, _ = parsed
            hour = time[:2]
            hour_counts[hour] = hour_counts.get(hour, 0) + 1

    if not hour_counts:
        return None

    return max(hour_counts, key=hour_counts.get)
```

```python
class Person:
    population = 0

    def __init__(self, name,age):
        self.name = name
        self.age = age
        Person.population += 1

    @classmethod
    def init_from_dict(cls,d):
        return cls(d['name'],d['age'])
    
    @classmethod
    def count_population(cls):
        return cls.population
    
    def introduce_myself(self):
        print(f'hello, my name is {self.name}')

    @staticmethod
    def say_hi():
         print('hi!')

a1 = Person.init_from_dict({"name": "a", "age":10})
a2 = Person('b',20)
a3 = Person('c',20)

print(Person.count_population())
a1.say_hi()
```
