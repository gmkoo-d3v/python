```python
f = open('./test.txt','r')
line_num = 1
line_data = f.readline()
while line_data:
    print(f"{line_num}: {line_data}", end = "")
    line_data = f.readline()
    line_num += 1
```

```python
f = open('./test.txt', 'r')
lines = f.readline()
for line_num, line_data in enumerate(lines):
    print(f"{line_num+1} {line_data}", end="")
f.close()
```

```python
with open('./newfile.txt', 'w', encoding='utf-8') as f:
    
    for i in range(1, 11):
        data = f"{i}번째 줄입니다. \n"
        f.write(data)
    print("파일 쓰기 완료!")
```
