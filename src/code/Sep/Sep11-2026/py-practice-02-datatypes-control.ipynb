### 문제 1-1. 파일 크기를 사람이 읽는 단위로 · 5분

`1234567` 바이트는 몇 MB 몇 KB 몇 B 일까요? 1KB는 1024B, 1MB는 1024KB 입니다.

**목표 변수**: `size` — `(MB, KB, B)` 순서의 튜플

<details>
<summary>힌트 보기</summary>

큰 단위부터 `//` 로 떼어내고, 남은 것(`%`)을 다음 단위로 넘깁니다.

</details>

```python
total_bytes = 1234567
KB = 1024
MB = 1024 * 1024

# 여기에 코드를 작성하세요
mb ,rest = divmod(total_bytes, MB)

kb, b = divmod(rest, KB)


size = (mb, kb, b)
print(size)
```

### 문제 1-2. 음수에서 달라지는 것들 · 5분

파이썬의 `//` 와 `int()` 는 음수에서 서로 다르게 동작합니다. 직접 확인하세요.

**목표 변수**: `q`, `r`, `truncated`, `rounded`

<details>
<summary>힌트 보기</summary>

`-7 // 2` 는 -3.5 를 **더 작은 쪽**으로 보냅니다. `int(-3.7)` 은 **0 쪽**으로 당깁니다.

</details>

```python
# 1) -7 을 2로 나눈 몫
q = (-7) // 2

# 2) -7 을 2로 나눈 나머지
r = (-7) % 2
# 3) -3.7 의 소수점을 잘라낸 값
truncated = int(-3.7)

# 4) -2.5 를 반올림한 값
rounded = round(-2.5)

print(q, r, truncated, rounded)
```

**채점 1-2** — 아래 셀을 실행하세요.

```python
assert q == -4, "// 는 '작은 쪽으로' 내림합니다 → -4 / 실제: " + repr(q)
assert r == 1, "파이썬의 나머지는 항상 나누는 수의 부호를 따릅니다 → 1 / 실제: " + repr(r)
assert truncated == -3, "int() 는 0 쪽으로 잘라냅니다 → -3 / 실제: " + repr(truncated)
assert rounded == -2, "가까운 짝수로 갑니다 → -2 / 실제: " + repr(rounded)
print("통과 — // 는 내림, int() 는 0 쪽으로 자름. 음수에서 갈립니다")
```

### 문제 1-3. 색상 코드 읽기 · 5분

웹 색상 `#1A2B3C` 는 빨강·초록·파랑을 각각 16진수 두 자리로 적은 것입니다. 10진수로 바꾸세요.

`ones` 는 다른 문제입니다 — `200` 을 2진수로 썼을 때 `1` 이 몇 개인지 세세요.

**목표 변수**: `r`, `g`, `b`, `ones`

<details>
<summary>힌트 보기</summary>

`int("1A", 16)` 처럼 진법을 두 번째 인자로 넘깁니다. 문자열도 `.count()` 를 씁니다.

</details>

```python
color = "1A2B3C"


# 여기에 코드를 작성하세요
r = int(color[0:2], 16)
g = int(color[2:4], 16)
b = int(color[4:], 16)

ones = bin(200).count('1')  # bin(200) 안에 '1' 이 몇 개?
print(r, g, b, ones)
```

### 문제 1-4. 단리와 복리 · 6분

원금 100만원을 연 이율 3%로 3년 둡니다.

- `simple` — 단리. 매년 원금의 3%씩만 붙습니다
- `compound` — 복리. 이자에 다시 이자가 붙습니다 (원금 × 1.03의 3제곱)

둘 다 **원 단위 정수**로 만드세요 (소수점 버림).

**목표 변수**: `simple`, `compound`, `gap`

<details>
<summary>힌트 보기</summary>

거듭제곱은 `**` 입니다. `1.03 ** 3` 처럼 씁니다.

</details>

```python
principal = 1000000
rate = 0.03
years = 3

# 여기에 코드를 작성하세요
simple = int(principal + principal * rate * years)
compound = int(principal * (1 + rate) ** years)
gap = compound-simple  # 복리가 단리보다 얼마나 더 버는가

print(simple, compound, gap)
```

**채점 1-4** — 아래 셀을 실행하세요.

```python
assert simple == 1090000, "100만 + 3만x3 = 109만 / 실제: " + repr(simple)
assert compound == 1092727, "100만 x 1.03^3 = 1,092,727 / 실제: " + repr(compound)
assert gap == 2727, "실제: " + repr(gap)
print("통과")
```

파이썬은 `True`/`False` 말고도 여러 값을 참·거짓으로 해석합니다. 어떤 것이 거짓인지 확인하세요.

**목표 변수**: `t1`, `t2`, `t3`, `t4`, `in_range`

<details>
<summary>힌트 보기</summary>

비교는 `0 <= score <= 100` 처럼 이어서 쓸 수 있습니다. 수학 그대로입니다.

```python
score = 85

t1 = bool('')  # bool("")      빈 문자열
t2 = bool('0')  # bool("0")     '0' 이라는 글자 하나
t3 = bool(0)  # bool(0)       숫자 0
t4 = bool([0]) # bool([0])     0 하나가 든 리스트

in_range = 0 < score <= 100  # score 가 0 이상 100 이하인가 (한 줄로)

print(t1, t2, t3, t4, in_range)
```

### 문제 2-2. 메뉴판 줄 맞추기 · 7분

가게 메뉴판을 텍스트로 만듭니다. 정렬 메서드를 씁니다.

- `title_line` — `"메뉴"` 를 너비 20으로 **가운데 정렬**, 남는 자리는 `=` 로 채우기
- `price_line` — 상품명을 너비 12로 **왼쪽 정렬**하고 남는 자리를 `.` 으로 채운 뒤, 가격을 너비 8로 **오른쪽 정렬**해서 붙이기

**목표 변수**: `title_line`, `price_line`

<details>
<summary>힌트 보기</summary>

`center(너비, 채울글자)`·`ljust(너비, 채울글자)`·`rjust(너비)` 입니다. 채울 글자를 생략하면 공백입니다.

```python
name = "아메리카노"
price_text = "4500원"
title = '메뉴'
# 여기에 코드를 작성하세요
title_line = f'{title.center(20,"=")}'
price_line = f'{name.ljust(12,".")}{price_text.rjust(8)}'

print(title_line)
print(price_line)
```

### 문제 2-3. find 는 못 찾으면 -1 · 5분

`find` 는 두 번째 인자로 **어디서부터 찾을지**를 줄 수 있습니다. 없는 것을 찾으면 에러가 아니라 `-1` 을 줍니다.

**목표 변수**: `first`, `second`, `count`, `not_found`

<details>
<summary>힌트 보기</summary>

`text.find("python", 시작위치)` 로 두 번째를 찾습니다.

</details>

```python
text = "python is fun, python is fast"

# 1) 첫 번째 python 의 위치
first = text.find('python',0)
# 2) 그 다음 python 의 위치 (1번 위치 다음부터 찾기)
second = text.find('python', first + 1)
# 3) python 이 몇 번 나오는지
count = text.count("python")
# 4) 없는 단어 "java" 를 찾으면?
not_found = text.find("java")

print(first, second, count, not_found)
```

**채점 2-3** — 아래 셀을 실행하세요.

```python
assert first == 0, "실제: " + repr(first)
assert second == 15, "실제: " + repr(second)
assert count == 2, "실제: " + repr(count)
assert not_found == -1, "find 는 못 찾으면 -1 을 줍니다 / 실제: " + repr(not_found)
print("통과 — index() 는 에러를 내지만 find() 는 -1 을 줍니다")
```

**목표 변수**: `every2`, `backwards`, `middle`, `last3`, `trimmed`, `first_half`

<summary>힌트 보기</summary>

`[시작:끝:간격]` 입니다. 음수 인덱스는 뒤에서부터, 음수 간격은 역순입니다.

</details>

```python
text = "abcdefghij"

every2 = text[::2]  # 한 글자 건너뛰며
backwards = text[::-1]  # 통째로 뒤집기
middle = text[3:7]   # 인덱스 3부터 6까지 (d~g)
last3 = text[-3::]  # 마지막 세 글자
trimmed = text[1:-1]  # 맨 앞과 맨 뒤 한 글자씩 뺀 것
first_half = text[:5:] # 앞 절반

print(every2, backwards, middle, last3, trimmed, first_half)
```

</details>

```python
board = [[0, 0], [0, 0]]
shallow = board[:]     # 겉만 복사했다

board[0][0] = 9        # 안쪽 리스트의 값을 바꾼다

# shallow[0][0] 은 0 일까 9 일까?
shallow_00 = shallow[0][0]
is_same_outer = shallow is board  # shallow 와 board 가 같은 물건인가
is_same_inner = shallow[0] is board[0]  # shallow[0] 과 board[0] 이 같은 물건인가

print(shallow)
print(shallow_00, is_same_outer, is_same_inner)
```

## Part 5 · 반복문 · 35분

while 로 숫자를 분해하고, 중첩 반복으로 표를 훑고, for-else 로 '못 찾음'을 처리합니다.

### 문제 5-1. 숫자를 자릿수로 분해하기 · 6분

`while` 과 `% 10`, `// 10` 만으로 숫자를 한 자리씩 뜯어냅니다. 아주 자주 쓰는 기법입니다.

`9305` 에서 자릿수의 합·개수·거꾸로 읽은 수를 구하세요.

**목표 변수**: `digit_sum`, `digit_count`, `reversed_n`

<details>
<summary>힌트 보기</summary>

`rest % 10` 이 마지막 자리, `rest // 10` 이 그것을 떼어낸 나머지입니다. 거꾸로 만들 때는 `기존 * 10 + 새 자리`.

</details>

```python
n = 9305

rest = n           # n 은 그대로 두고 복사본을 부순다
digit_sum = 0
digit_count = 0
reversed_n = 0

while rest > 0:
    # 1) 마지막 자리 숫자 꺼내기 (% 10)
    # 2) 합·개수에 반영
    # 3) reversed_n 을 한 칸 밀고 그 숫자 붙이기
    # 4) rest 를 한 자리 줄이기 (// 10)
    # break  # ← 이 줄을 지우고 반복할 일을 직접 작성하세요
    digit = rest % 10

    digit_sum += digit
    digit_count += 1

    reversed_n = reversed_n * 10 + digit

    rest //= 10

print(digit_sum, digit_count, reversed_n)
```

**채점 5-1** — 아래 셀을 실행하세요.

```python
assert digit_sum == 17, "9+3+0+5 = 17 / 실제: " + repr(digit_sum)
assert digit_count == 4, "실제: " + repr(digit_count)
assert reversed_n == 5039, "9305 를 거꾸로 읽으면 5039 / 실제: " + repr(reversed_n)
print("통과")
```

`powers` 는 `2 ** i` 를 담아도 되고, 값을 계속 2배 해나가도 됩니다.

```python
# 1) 5! = 1x2x3x4x5
factorial = 1

for i in range(1,6):
    factorial *= i

# 2) 2의 0제곱부터 4제곱까지 담은 리스트 [1, 2, 4, 8, 16]
powers = []

for i in range(5):
    powers.append(2 ** i)


print(factorial, powers)
```

**채점 5-2** — 아래 셀을 실행하세요.

```python
assert factorial == 120, "5! = 120 / 실제: " + repr(factorial)
assert powers == [1, 2, 4, 8, 16], "실제: " + repr(powers)
print("통과")
```

### 문제 5-4. 중첩 반복으로 세기 · 7분

2단부터 5단까지 구구단을 훑으면서 **곱이 짝수인 경우만** 셉니다. 홀수는 `continue` 로 건너뛰세요.

<details>
<summary>힌트 보기</summary>

짝수 단은 전부 짝수가 나오고, 홀수 단은 곱하는 수가 짝수일 때만 짝수가 됩니다.

```python
even_count = 0

for dan in range(2, 6):
    for i in range(1, 10):
        # 여기에 코드를 작성하세요 (continue 를 쓰세요)
        # ← 이 줄을 지우고 여기에 작성하세요
        result = dan * i

        if result % 2 != 0:
            continue

        even_count += 1


print(even_count)
```

### 문제 5-5. for-else 로 '못 찾음' 처리하기 · 5분

두 사람을 명단에서 찾습니다. 찾으면 `"찾음"`, 끝까지 없으면 `"없음"` 을 결과에 담으세요.

**목표 변수**: `results`

<details>
<summary>힌트 보기</summary>

`else` 는 **안쪽 `for`** 와 같은 들여쓰기 높이에 붙입니다.

</details>

```python
names = ["하늘", "민수", "지연"]
targets = ["민수", "도윤"]
results = []

for target in targets:
    for name in names:
        # 찾으면 결과에 "찾음" 을 넣고 break
         # ← 이 줄을 지우고 여기에 작성하세요
        if target == name:
            results.append("찾음")
            break
    else:
        # break 없이 for 문이 끝난 경우
        results.append("없음")

    # 여기에 else: 를 붙여 "없음" 을 넣으세요


print(results)
```

**채점 5-5** — 아래 셀을 실행하세요.

```python
assert results == ["찾음", "없음"], "민수는 있고 도윤은 없습니다 / 실제: " + repr(results)
print("통과 — else 는 break 없이 끝났을 때만 실행됩니다")
```

### 문제 5-6. 카테고리별 매출 집계 · 5분

판매 기록을 카테고리별로 합치고, 가장 많이 판 카테고리를 찾으세요.

**목표 변수**: `totals`, `best_category`

<details>
<summary>힌트 보기</summary>

`for category, amount in sales` 처럼 반복하면서 바로 튜플을 풀 수 있습니다.

</details>

```python
sales = [("과일", 3000), ("채소", 2000), ("과일", 1500), ("음료", 800), ("채소", 1200)]

totals = {}
for category, amount in sales:
    # get 을 쓰면 한 줄입니다
    # ← 이 줄을 지우고 여기에 작성하세요

    totals[category] = totals.get(category, 0) + amount

best_category = ""
best_amount = 0
for category, amount in totals.items():
    # 지금까지 최고보다 크면 갱신
    # ← 이 줄을 지우고 여기에 작성하세요
    if amount > best_amount:
        best_amount = amount
        best_category = category


print(totals, best_category)
```

**채점 5-6** — 아래 셀을 실행하세요.

```python
assert totals == {"과일": 4500, "채소": 3200, "음료": 800}, "실제: " + repr(totals)
assert best_category == "과일", "실제: " + repr(best_category)
print("통과")
```

출석부와 텍스트 통계. 한 번의 반복 안에서 여러 가지를 동시에 계산합니다.

### 문제 6-1. 출석부 집계 · 8분

출결 기록을 한 번의 반복으로 네 가지로 정리하세요.

| 출결 | 점수 |
|---|---|
| 출석 | 1.0 |
| 지각 | 0.5 |
| 결석 | 0 |

**목표 변수**: `counts`, `total_score`, `absent_names`, `rate`
(`rate` — 출석 점수 합을 인원수로 나눈 백분율, 소수점 1자리)

<details>
<summary>힌트 보기</summary>

반복 하나 안에서 세 가지를 동시에 합니다. 누적 변수는 반복 **밖에서** 먼저 만들어 둡니다.

</details>

```python
records = [("하늘", "출석"), ("민수", "지각"), ("지연", "결석"), ("도윤", "출석"), ("서준", "지각")]

counts = {}
total_score = 0
absent_names = []

for name, status in records:
    # 1) counts 에 출결 종류별 인원 세기
    counts[status] = counts.get(status, 0) + 1
    # 2) 점수를 total_score 에 누적
    if status == "출석":
        total_score += 1
    elif status == "지각":
        total_score += 0.5
    else:
        total_score += 0
    # 3) 결석이면 absent_names 에 이름 추가
    # ← 이 줄을 지우고 여기에 작성하세요
    if status == "결석":
        absent_names.append(name)


rate = total_score / len(records) * 100

print(counts, total_score, absent_names, rate)
```

**채점 6-1** — 아래 셀을 실행하세요.

```python
assert counts == {"출석": 2, "지각": 2, "결석": 1}, "실제: " + repr(counts)
assert total_score == 3.0, "1 + 0.5 + 0 + 1 + 0.5 = 3.0 / 실제: " + repr(total_score)
assert absent_names == ["지연"], "실제: " + repr(absent_names)
assert rate == 60.0, "3.0 / 5 x 100 = 60.0 / 실제: " + repr(rate)
print("통과")
```

### 문제 6-2. 텍스트 통계 리포트 · 7분

문장을 분석해 다섯 가지를 구하세요. `max()` 를 쓰지 말고 반복으로 직접 찾습니다.

**목표 변수**: `word_count`, `unique_count`, `the_count`, `longest`, `avg_len`
(`avg_len` — 단어 길이의 평균, 소수점 2자리)

<details>
<summary>힌트 보기</summary>

`>` 로 비교하면 **먼저 나온 것**이 남고, `>=` 로 비교하면 나중 것이 남습니다.

</details>

```python
text = "the quick brown fox jumps over the lazy dog the end"
words = text.split()

word_count = len(words)
unique_count = len(set(words))
the_count = words.count('the')

longest = ""
length_sum = 0
for word in words:
    # 1) 지금까지 본 것보다 길면 longest 갱신
    if len(word) > len(longest):
        longest = word
    # 2) 길이를 length_sum 에 누적
    # ← 이 줄을 지우고 여기에 작성하세요
    length_sum += len(word)


avg_len = round(length_sum / word_count, 2)

print(word_count, unique_count, the_count, longest, avg_len)
```

**채점 6-2** — 아래 셀을 실행하세요.

```python
assert word_count == 11, "실제: " + repr(word_count)
assert unique_count == 9, "the 가 세 번 나오므로 중복을 빼면 9 / 실제: " + repr(unique_count)
assert the_count == 3, "실제: " + repr(the_count)
assert longest == "quick", "quick/brown/jumps 가 모두 5글자인데, 먼저 나온 것이 남습니다 / 실제: " + repr(longest)
assert avg_len == 3.73, "41 / 11 = 3.727... / 실제: " + repr(avg_len)
print("통과")
```

### 문제 ★-3. 표에서 최댓값의 위치 찾기 · 10분

중첩 리스트에서 가장 큰 값과 **그 위치(행, 열)** 를 찾으세요.

**목표 변수**: `max_value`, `max_row`, `max_col`

<details>
<summary>힌트 보기</summary>

행 인덱스로 한 번, 그 행의 열 인덱스로 또 한 번 돕니다. 값보다 **위치**를 기억해야 합니다.

</details>

```python
matrix = [
    [3, 7, 1],
    [9, 2, 8],
    [4, 6, 5],
]

max_value = matrix[0][0]
max_row = 0
max_col = 0

# 여기에 코드를 작성하세요 (중첩 반복)
for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if matrix[i][j] > max_value:
            max_value = matrix[i][j]
            max_row = i
            max_col = j

print(max_value, max_row, max_col)
```

**채점 ★-3** — 아래 셀을 실행하세요.

```python
assert max_value == 9, "실제: " + repr(max_value)
assert max_row == 1 and max_col == 0, "9 는 2행 1열, 인덱스로는 (1, 0) / 실제: " + repr((max_row, max_col))
print("통과")
```
