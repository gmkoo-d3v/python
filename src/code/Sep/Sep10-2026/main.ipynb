```python
numbers = list(range(0,50,7))
print(numbers)
odd = []; even = []; zero = []
for nums in numbers:
    if nums == 0:
            zero.append(nums)
    elif nums % 2 == 0:
        even.append(nums)    
    else:
        odd.append(nums)
print("짝수:",even, "홀수",odd, "zero",zero)
```

```python
numbers = list(range(0,12,1))
print(numbers)
odd = []; even = []; zero = []; third = []
for nums in numbers:
    if nums == 0:
        zero.append(nums)
    elif nums %3 == 0:
        odd.append(nums)
        if nums % 3 == 0:   
            third.append(nums)
        if nums %2 == 0:
            even.append(nums) 
    elif nums % 2 == 0:
        even.append(nums)    
    else:
        odd.append(nums)
print("짝수:",even, "홀수",odd, "zero",zero, "3배수",third)
```

```python
# names 와 scores 가 주어집니다. 직접 시험해보려면 아래 두 줄의 # 을 지우세요.
names = ["민수", "영희", "지우"]
scores = {"민수": 88, "영희": 72, "지우": 91}

# 1단계 — 철수를 명단과 점수에 추가
names.append("철수"); scores["철수"]=78

# 2단계 — 영희를 95점으로
scores["영희"]=95

# 3단계 — 집계
average = sum(scores.values())/len(scores)
top = max(scores.values())
passed = []
for nam in scores:
    if scores[nam]>=80:
        passed.append(scores[nam])

print("평균", average, "/ 최고", top, "/ 통과", passed)
```

### 보조제어문
- continue
- break
- pass

```python
ind = 0
while True:
    ind += 1
    print(ind,"be if")
    if ind <= 3:
        print(ind, "if")
        pass
    elif ind < 10:
        print(ind,'before con')
        continue
    else:
        print(ind, 'else')
        break
    print("here!")
print("out")    
```
