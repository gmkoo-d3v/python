# 팩토리얼 예제

```python
# 반복적으로 구현한 n!
def factorial_iterartive(n):
    result = 1
    # 1 부터 n 까지의 수를 차례대로 곱하기
    for i in range(1, n + 1):
        result  *= i
    return result

# 재귀적으로 구현한 n!
def factorial_recursive(n):
    if n <= 1: # ndl 1 이하인 경우 1을 반환
        return 1
    # n!= n * (n-1)!를 그대로 코드로 작성하기
    return n * factorial_recursive(n - 1)

# 각각의 방식으로 구현한 n!  출력 (n = 5)
print(f'반복적으로 구현:{factorial_iterartive(5)}') 
print(f'반복적으로 구현:{factorial_recursive(5)}') 
```

# 애너그램 판별기
</br>두 개의 단어를 입력받아, 두 단어가 애너그램 관계인지 확인하는 함수를 작성하시오.
</br>애너그램: 동일한 문자를 다른 순서로 나열한 단어. 대소문자 무시, 공백제거
</br>결과는 True or False
</br>example: listen vs silent -> True
</br>example: hello vs world -> False
</br>example: dog vs god -> True

```python
def is_anagram(word1, word2):
    w1 = sorted(word1.replace(" ","").lower())
    w2 = sorted(word2.replace(" ","").lower())
    return w1 == w2

def main():
    word1 = input('첫 번째 단어를 입력하세요: ')
    word2 = input('두 번째 단어를 입력하세요: ')
    print('애너그램 여부:', is_anagram(word1, word2))

main()
```

# palindrome_stack

```python
def is_palindrom_stack(word):
    # 공백 제거 및 소문자 변환
    cleaned = word.replace(" ", "").lower()

    stack = []

    # 스텍에 모든 글자 push(넣기)
    for char in cleaned:
        stack.append(char)

    # 스텍에서 하나씩 pop(꺼내기)하며 원래 글자와 비교
    for char in cleaned:
        if char != stack.pop():
            return False
    return  True

def is_palindrom(word):
    cleaned = word.replace(" ", "").lower()
    return cleaned == "".join(reversed(cleaned))

# test
word = input("단어를 입력하세요")
print("stack 회문 여부:", is_palindrom_stack(word))
print("회문 여부:", is_palindrom(word))
```

```python
def solution(prices):
    n = len(prices)
    answer = [0] * n
    stack = []  # 가격이 떨어지지 않은 시점의 '인덱스'를 저장

    for i in range(n):
        # 현재 가격이 스택 top의 가격보다 낮다면 → 가격이 떨어진 것
        while stack and prices[stack[-1]] > prices[i]:
            j = stack.pop()
            answer[j] = i - j

        stack.append(i)

    # 끝까지 가격이 떨어지지 않은 시점들 처리
    while stack:
        j = stack.pop()
        answer[j] = (n - 1) - j

    return answer
print("stack:", stack)
print('answer:', answer)
```

# 문자열 압축

```python
def compress(word):
    answer = ''
    count = 1

    for i in range(len(word) - 1):
        if word[i] == word[i + 1]:
            count += 1
        else:
            answer += word[i]
            answer += str(count)
            count = 1

    answer += word[-1]
    answer += str(count)

    return answer

print(compress(input()))
```

```python
def find_longest_sentence():
    line = input("문장들을 '.'로 구분하여 입력하세요: ")
    sentences = line.split('.')
    
    longest = ''
    for s in sentences:
        s = s.strip()
        if len(s) > len(longest):
            longest = s
    print("가장 긴 문장:", longest)    



find_longest_sentence()
```

```python
# 유틸리티 메서드 구현에 자주 사용되는 static method

class StringUtils:
    @staticmethod
    def toCamelcase(text):
        words = iter(text.split("_"))
        return next(words) + "".join(i.title() for i in words)

    @staticmethod
    def toSnakecase(text):
        letters = ["_"+ i.lower() if i.isupper() else i for i in text]
        return "".join(letters.lstrip("_"))
```

```python
from abc import *

class Character(metaclass = ABCMeta):
    def __init__(self):
        self.hp = 100
        self.attack_power = 20

    def attack(self, other, attack_kind):
        other.get_damage(self.attack_power, attack_kind)

    @abstractmethod
    def get_damage(self, attack_power, attack_kind):
        pass

class Player(Character):
    def get_damage(self, attack_power, attack_kind):
        self.hp -= attack_power

char1 = Player()

class Player2(Character):
    pass

char2 = Player2()
# ---> 24 char2 = Player2()

# TypeError: Can't instantiate abstract class Player2 without an implementation for abstract method 'get_damage'
```

```python
from abc import ABC, abstractmethod


class Unit(ABC):

    def __init__(self, name):
        self.name = name
        self.hp = 100
        self.attack_power = 20

    @abstractmethod
    def weapon(self, attack_power, attack_kind):
        pass

    @abstractmethod
    def move(self, area):
        pass


class Marine(Unit):

    def move(self, area):
        print(f'{self.name}이 {area}로 걸어서 이동')

    def weapon(self, attack_power, attack_kind):
        print(f'{attack_kind}로 {attack_power}의 데미지')


class Plane(Unit):

    def move(self, area):
        print(f'{self.name}이 {area}로 날아서 이동')

    def weapon(self, attack_power, attack_kind):
        print(f'{attack_kind}로 {attack_power}의 데미지')


character1 = Marine("마린")
character2 = Plane("비행기")

character1.move("전방")
character2.move("전방")

character1.weapon(30, "소총")
character2.weapon(50, "레이저")
```
