### 문제 1-6. 반올림의 함정 · 4분

파이썬의 `round()` 는 우리가 배운 '5면 올림'과 다르게 동작합니다. 직접 확인하세요.

**목표 변수**: `r1`, `r2`, `r3`, `fixed`

<details>
<summary>힌트 보기</summary>

`int()` 는 소수점 아래를 **버립니다**. 미리 0.5를 더해두면?

</details>

````python
# 1) round(2.5) 의 결과
r1 = round(2.5)

# 2) round(3.5) 의 결과
r2 = round(3.5)

# 3) round(2.675, 2) 의 결과
r3 = round(2.675, 2)

# 4) 2.5 를 '항상 올림'으로 반올림해서 3 을 만드세요 (round 를 쓰지 않고)
fixed = int(2.5+0.5)

print(r1, r2, r3, fixed)
````

### 문제 2-6. 파일명 정리 파이프라인 · 5분

지저분한 파일명을 단계별로 정리합니다. **각 단계를 따로따로 변수에 담으세요** — 무엇이 바뀌는지 눈으로 보기 위해서입니다.

**목표 변수**: `step1`, `step2`, `step3`, `name_part`, `ext`

<details>
<summary>힌트 보기</summary>

메서드는 원본을 바꾸지 않고 **새 문자열을 돌려줍니다**. 그래서 결과를 변수에 받아야 합니다.

</details>

````python
raw = "   ImSi-Data_2026.CSV   "

# 1) 양끝 공백 제거
step1 = raw.strip()
# 2) 전부 소문자로
step2 = step1.lower()
# 3) 하이픈(-)을 밑줄(_)로
step3 = step2.replace('-',"_")
# 4) 확장자를 뺀 이름 / 확장자
# name_part = step3.split('.')[0]
# ext = step3.split('.')[1]

splited = step3.split(".")
if len(splited) == 2:
    name_part, ext = splited
else:
    ext = splited[-1]
    name_part = ".".join(splited[:-1])

print(repr(step1)); print(step2); print(step3)
print(name_part, ext)
````

**채점 2-6** — 아래 셀을 실행하세요.

</details>

### 문제 6-1. 성적 처리 시스템 · 8분

점수 딕셔너리를 받아 학점·평균·최고점자·과락자를 한 번에 구합니다.

| 점수 | 학점 |
|---|---|
| 90 이상 | A |
| 80 이상 | B |
| 70 이상 | C |
| 60 이상 | D |
| 60 미만 | F |

**목표 변수**: `grades`, `average`, `top_name`, `fail_names`

<details>
<summary>힌트 보기</summary>

반복 한 번 안에서 네 가지를 동시에 처리합니다. 누적 변수를 반복 **밖에서** 먼저 만들어 두는 것이 핵심입니다.

</details>

````python
scores = {"하늘": 92, "민수": 78, "지연": 55, "도윤": 88, "서준": 61}

grades = {}
fail_names = []
top_name = ""
top_score = 0
total = 0
average = 0



for name, score in scores.items():
    # 1) 학점 매기기

    # 2) 과락(60 미만) 명단

    # 3) 최고점자

    # 4) 합계 누적

    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = 'F'
    grades[name] = grade
    total += score
    if score > top_score:
        top_score = score
        top_name = name
    if score < 60:
        fail_names.append(name)
average = round(total/len(scores),1)


print(grades)
print(average, top_name, fail_names)
````

**채점 6-1** — 아래 셀을 실행하세요.

````python
assert grades == {"하늘": "A", "민수": "C", "지연": "F", "도윤": "B", "서준": "D"}, "실제: " + repr(grades)
assert average == 74.8, "(92+78+55+88+61)/5 = 74.8 / 실제: " + repr(average)
assert top_name == "하늘", "실제: " + repr(top_name)
assert fail_names == ["지연"], "실제: " + repr(fail_names)
print("통과")
````

- `total` — 최종 결제액
- `lines` — 품목당 한 줄. 이름은 너비 10 왼쪽 정렬, 수량은 `x2` 형태로 너비 4, 금액은 `9000원` 형태로 너비 8 오른쪽 정렬

<details>
<summary>힌트 보기</summary>

`for name, price, qty in cart` 처럼 **반복하면서 바로 언패킹**할 수 있습니다.

</details>

````python
cart = [
    ("아메리카노", 4500, 2),
    ("샌드위치", 6800, 1),
    ("쿠키", 2500, 3),
]

subtotal = 0
lines = []

for name, price, qty in cart:
    # 여기에 코드를 작성하세요
    ...  # ← 이 줄을 지우고 여기에 작성하세요
    amount = price * qty
    subtotal += amount
    lines.append(name.ljust(15)+("x"+str(qty).ljust(4)+(str(amount)+"원")))



# 소계가 20000 이상이면 10% 할인
dc = 20000
discount = int(subtotal * .1)
if subtotal >= dc:
    total = subtotal-discount

for line in lines:
    print(line)
print(subtotal, discount, total)
````
