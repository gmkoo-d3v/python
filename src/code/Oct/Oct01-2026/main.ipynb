```python
from idlelib.debugobj import dispatch

import pandas as pd
import numpy as np
```

```python

df1 = pd.DataFrame(
    {'Favorite Food': ['Alfredo Pasta', "Bobba", 'Croissants', "danish Rolls", 'Edamame', 'Fish', "Ice Cream"]},
    index=pd.Index(['Alfred', 'Bob', 'Cris', 'Daniel', 'Edward', 'Fitx', 'Ivan'], name='Name'))
df2 = pd.DataFrame(
    {'Favorite Country': ['Albania', 'Croatia', 'Denmark', 'Fiji', 'Germany', 'Haiti']},
    index=pd.Index(['Alfred', 'Cris', 'Daniel', 'Edward', 'Fitx', 'Ivan'], name='Name'))


display(df1.join(df2, how= 'outer'))  # 합집합
display(df1.join(df2, how= 'inner')) #교집합
```

```python
display(df1.join(df2))
```

```python
display(df1.join(df2, how= 'right'))
```

```python
# pd.DataFreame : merge (join 보다 더 똑똑한 개념 = merge)
df1 = df1.reset_index()
display(df1)
df2 = df2.reset_index()
display(df2)

# 교집합이 'inner'가  default 값 'how' = 'outer', 'left', 'right' 선택 가능
display(pd.merge(df1,df2))
display(df1.merge(df2))

#
df3 = pd.DataFrame({'Favorite Course':["Criminomogy", "Design", "Fitness"],
                    'Favorite Country':['Croatia','Denmark','Fiji']})
display(df3)
display(pd.merge(df2, df3, left_on='Favorite Country', right_on='Favorite Country'))
```

```python
df4 = pd.DataFrame({'Name':['Cris', 'Daniel','Fitz','Peter'],
                    'Favorite Course':["Criminomogy", "Design", "Fitness",'Yoga'],
                    'Favorite Country':['Croatia','Denmark','Fiji','Croatia'],
                    'Favorite Food':['Croissants','Danish Rolls', 'Fish', 'Meat']})
display(df4)
display(df2)

df2.merge(df4, on = ['Favorite Country'])
```

```python
df = pd.DataFrame({'foo':['one','one','one','two','two','two'],
                   'bar':['A','B','C','A','B','C'],
                   'baz':[1,2,3,4,5,6],
                   'zoo':['x','y','z','q','w','t']})
display(df)
display(df.pivot(index='foo', columns='bar', values='baz'))

df3 = pd.DataFrame({'first':['John', 'Mary'],
                    'last':['Doe', 'Bo'],
                    'height':[5.5, 6.0],
                    'weight':[130,150]})
display(df3)
display(df3.melt(id_vars=['first', 'last']))
```

```python
index = pd.MultiIndex.from_tuples([('bar','one'),('bar','two'),
                                   ('baz','one'),('baz','Two')],
                                  name = ['first', 'second'])
df2 = pd.DataFrame([[1,2],[3,4],[5,6],[7,8]], index=index, columns=['A','B'])
display(df2)

stacked = df2.stack()
display(stacked)
print("-"*10)
print(stacked.index.names) # fist, second, none output
print("-"*10)

display(stacked.unstack())

print(stacked.unstack().index.names)
display(stacked.unstack(1)) # or stacked.unstack('second')
display(stacked.unstack(2)) # or stacked.unstack('first')
```
